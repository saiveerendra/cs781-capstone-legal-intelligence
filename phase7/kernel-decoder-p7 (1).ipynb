```python
class Node:
    def __init__(self,value, previous, next, index):
        self.value=value
        self.previous=previous
        self.next=next
        self.index=index


class LinkedArray:
    def __init__(self, items):
        self._array=[None for _ in range(len(items))]
        previous=None
        for i in range(len(items)):
            node=Node(items[i], previous, None, i)
            if previous is not None:
                node.previous=previous
                previous.next=node
            self._array[i]=node
            previous=node

    def get_previous_index(self, index):
        if self._array[index]==None:
            raise ValueError(f"No item at index {index}")
        if self._array[index].previous == None:
            return None
        return self._array[index].previous.index

    def get_next_index(self, index):
        if self._array[index]==None:
            raise ValueError(f"No item at index {index}")
        if self._array[index].next == None:
            return None
        return self._array[index].next.index

    def get_by_index(self, index):
        if self._array[index]==None:
            raise ValueError(f"No item at index {index}")
        if self._array[index] == None:
            return None
        return self._array[index].value

    def get_second_next_index(self, index):
        if self._array[index] == None:
            raise ValueError(f"No item at index {index}")
        if self._array[index].next == None or self._array[index].next.next == None:
            return None
        return self._array[index].next.next.index

    def len(self):
        return len(self._array)
    def exists(self, index):
        return (
            0 <= index < len(self._array)
            and self._array[index] is not None
        )

    def replace_pair(self, index, new_item):
        if index > len(self._array)-2 or self._array[index] == None or self._array[index].next == None:
            raise ValueError("Invalid index ",index,new_item)
        self._array[index].value = new_item
        self._array[self._array[index].next.index] = None
        self._array[index].next = self._array[index].next.next
        if self._array[index].next is not None:
            self._array[index].next.previous = self._array[index]


class MaxPriorityMap:
    def __init__(self, heap_key, map_key):
        self._heap_key = heap_key
        self._map_key = map_key
        self._heap = []
        # Map of keys to indexes in the heap
        self._map = {}

    def get_max(self):
        if len(self._heap) == 0:
            raise ValueError("the heap is empty")
        return self._heap[0]

    def push(self, item):
        self._heap.append(item)
        map_key = self._map_key(item)
        if map_key in self._map:
            raise ValueError("Map key already exist")
        self._map[self._map_key(item)] = len(self._heap) - 1
        self._heapify_up(len(self._heap) - 1)

    def pop(self):
        if len(self._heap) == 0:
            raise ValueError("the heap is empty")
        self._swap(0, len(self._heap) - 1)
        item = self._heap.pop()
        map_key = self._map_key(item)
        self._map.pop(map_key)
        self._heapify_down(0)
        return item

    def contains(self, map_key):
        return map_key in self._map

    def len(self):
        if len(self._heap) != len(self._map):
            raise ValueError('Invalid state of the priority map')
        return len(self._heap)

    def delete_by_map_key(self, map_key):
        if map_key not in self._map:
            raise ValueError("Map key does not exist")
        heap_index = self._map[map_key]
        item = self._heap[heap_index]
        self._swap(heap_index, len(self._heap) - 1)
        self._heap.pop()
        self._map.pop(map_key)
        if (heap_index < len(self._heap)):
            self._heapify_down(heap_index)
            self._heapify_up(heap_index)
        return item

    def _heapify_up(self, i):
        while i > 0 and self._heap_key(self._heap[i]) > self._heap_key(self._heap[self._parent(i)]):
            self._swap(i, self._parent(i))
            i = self._parent(i)

    def _heapify_down(self, i):
        largest = i
        left = self._left_child(i)
        right = self._right_child(i)

        if left < len(self._heap) and self._heap_key(self._heap[left]) > self._heap_key(self._heap[largest]):
            largest = left

        if right < len(self._heap) and self._heap_key(self._heap[right]) > self._heap_key(self._heap[largest]):
            largest = right

        if largest != i:
            self._swap(i, largest)
            self._heapify_down(largest)

    def _parent(self, i):
        return (i - 1) // 2

    def _left_child(self, i):
        return 2 * i + 1

    def _right_child(self, i):
        return 2 * i + 2

    def _swap(self, i, j):
        self._map[self._map_key(self._heap[i])] = j
        self._map[self._map_key(self._heap[j])] = i
        self._heap[i], self._heap[j] = self._heap[j], self._heap[i]


class StatsEntry:
    def __init__(self, pair, positions):
        self.pair = pair
        self.positions = positions

class TokenizerTrainer:
        def __init__(self, tokens_list, itos, num_merges):
            self.tokens_list = tokens_list
            self.itos = itos
            self._num_merges = num_merges
            self.stoi={}

        def train(self, next_token):
            self._positions=[LinkedArray(tokens) for tokens in self.tokens_list]
            self.get_stats()
            merges_done=0
            print("requested merges",self._num_merges)
            while  merges_done<self._num_merges:
                if self._stats.len() == 0:
                   print("STOPPED: stats became empty")
                   print("Merges completed:", merges_done)
                   break
                merge_stat = self._stats.pop()
                (current_token , next_token) = self.update_itos_and_stoi(next_token, merge_stat)
                #iterate through posistions replace with merge pair
                for position in list(merge_stat.positions):

                    if position not in merge_stat.positions:
                        continue
                    input_index = position[0]
                    token_index = position[1]
                    # print("current_token",current_token)
                    if not self._positions[input_index].exists(token_index):
                        merge_stat.positions.discard(position)
                        continue
                    self.update_right_token(input_index, token_index, merge_stat, current_token)
                    self.update_left_token(input_index, token_index, merge_stat, current_token)
                    self._positions[input_index].replace_pair(token_index, current_token)
                    merge_stat.positions.remove((input_index, token_index))
                merges_done += 1
            return self.itos


        def update_right_token(self, input_index, token_index, merge_stat, current_token):
           position = self._positions[input_index]
           next_token_idx = position.get_next_index(token_index)
           right_token_idx = position.get_second_next_index(token_index)
           if right_token_idx == None:
               return
           pair = (merge_stat.pair[1], position.get_by_index(right_token_idx))
           self.remove_pair_from_stats(merge_stat, pair, input_index, next_token_idx)
           new_pair = (current_token, pair[1])
           self.add_position_to_pair(new_pair, input_index, token_index)

        def update_left_token(self, input_index, token_index, merge_stat, current_token):
           position = self._positions[input_index]
           left_token_idx = position.get_previous_index(token_index)
           if left_token_idx == None:
               return
           pair = ( position.get_by_index(left_token_idx), merge_stat.pair[0])
           self.remove_pair_from_stats(merge_stat, pair, input_index, left_token_idx)
           new_pair = (pair[0] , current_token)
           self.add_position_to_pair(new_pair, input_index, left_token_idx)

        def add_position_to_pair(self, new_pair, input_index, token_index):
            if self._stats.contains(new_pair):
                stat = self._stats.delete_by_map_key(new_pair)
            else:
                stat = StatsEntry(new_pair, set())

            stat.positions.add((input_index, token_index))
            self._stats.push(stat)


        def remove_pair_from_stats(self, merge_stat, pair, input_index, token_index):
           if pair == merge_stat.pair:
               merge_stat.positions.discard((input_index, token_index))
           else:
                if self._stats.contains(pair):
                    stat = self._stats.delete_by_map_key(pair)
                    stat.positions.discard((input_index, token_index))
                    if len(stat.positions) != 0:
                        self._stats.push(stat)


        def update_itos_and_stoi(self, next_token, merge_stat):
           merge_pair = self.itos[merge_stat.pair[0]] + self.itos[merge_stat.pair[1]]
           if merge_pair in self.stoi:
               current_token = self.stoi[merge_pair]
           else:
               current_token = next_token
               self.itos[current_token] = merge_pair
               self.stoi[merge_pair] = current_token
               next_token = current_token+1
           return  (current_token , next_token)

        def get_stats(self):
             self._stats = MaxPriorityMap(
                                heap_key=lambda item: (len(item.positions),tuple(-x for x in item.pair)),
                                map_key=lambda item: item.pair
                            )
             stats={}
             for  text_id in range(len(self.tokens_list)):
                 tokens = self.tokens_list[text_id]
                 for idx in range(len(tokens)-1):
                     pair = (tokens[idx], tokens[idx+1])
                     if pair not in stats:
                         stats[pair]=StatsEntry(pair, set())
                     stats[pair].positions.add((text_id, idx))
             for key in stats.keys():
                self._stats.push(stats[key])

class TokenNode:
    def __init__(self, byte_value, token, parent, children):
        self.parent = parent
        self.byte_value = byte_value
        self.token = token
        self.children = children

class ToTokensConverter:

    def __init__(self, itos):
        self.itos = itos
        self.unk_token=1
        self._init_token_tree()

    def to_tokens(self, strings):
        return [
            self._to_tokens(string)
            for string in strings
        ]

    def _to_tokens(self, string):

        words = string.split()
        tokens = []
        for word in words:
            data = word.encode("utf-8") + b"</w>"
            word_tokens = self._tokenize_word(data)
            tokens.extend(word_tokens)
        return tokens

    def _tokenize_word(self, data):

        tokens = []

        byte_index = 0

        while byte_index < len(data):

            current_nodes = self._token_roots

            last_node_with_token = None
            last_index_with_token = byte_index

            while byte_index < len(data):

                byte_value = data[byte_index]

                if byte_value in current_nodes:

                    node = current_nodes[byte_value]

                    if node.token is not None:
                        last_node_with_token = node
                        last_index_with_token = byte_index

                    current_nodes = node.children
                    byte_index += 1

                else:
                    break

            if last_node_with_token is None:
                tokens.append(self.unk_token)

                # Move forward by one byte
                byte_index += 1

            else:
                tokens.append(last_node_with_token.token)
                byte_index = last_index_with_token + 1

        return tokens

    def _init_token_tree(self):

        self._token_roots = {}

        for token, token_bytes in self.itos.items():

            if token < 4:
                continue

            current_nodes = self._token_roots
            last_node = None

            for byte_value in token_bytes:

                if byte_value in current_nodes:

                    last_node = current_nodes[byte_value]

                else:

                    last_node = TokenNode(
                        byte_value,
                        None,
                        last_node,
                        {}
                    )

                    current_nodes[byte_value] = last_node

                current_nodes = last_node.children

            if last_node.token is not None:
                raise KeyError("The token is already taken")

            last_node.token = token

class BPETokenizer:
    def __init__(self, num_merges):
        self._num_merges = num_merges
        self.itos = {}
    def get_vocab_size(self):
        return len(self.itos)
    def train(self, texts):
        self.create_base_vocab()

        tokens_list = self.get_ids_from_text(texts)
        # for tokens in tokens_list:
        #     print(tokens)
        trainer = TokenizerTrainer(
                        tokens_list,
                        self.itos,
                        self._num_merges
                    )

        trainer.train(261)

        return self.itos, tokens_list

    def encode(self, text):
        converter = ToTokensConverter(self.itos)
        return converter.to_tokens([text])[0]

    def decode(self, tokens):
        return (
                b"".join(
                    self.itos[token]
                    for token in tokens
                )
                .decode("utf-8", errors="replace")
                .replace("</w>", " ")
                .strip()
            )

    def create_base_vocab(self):
        self.itos = {
            0: b"<pad>",
            1: b"<unk>",
            2: b"<s>",
            3: b"</s>",
            260 : b"</w>"
        }
        for i in range(256):
            self.itos[i + 4] = bytes([i])

    def get_ids_from_text(self, texts):
        tokens_list=[]
        for text in texts:
            words = text.split()
            for word in words:
                word_bytes = [byte + 4 for byte in word.encode("utf-8")]
                word_bytes.append(260)
                tokens_list.append(word_bytes)
        return tokens_list
```

```python

def convert_text_to_ids(text):
    token_ids = converter.to_tokens([text])[0]
    if len(token_ids) == 0:
        return [UNK_ID]
    token_ids = [int(x) for x in token_ids]
    bad_ids = [x for x in token_ids if x < 0 or x >= VOCAB_SIZE]
    if bad_ids:
        raise ValueError(f"Invalid token ids: {bad_ids[:10]}")
    return token_ids

print("Vocabulary size:", VOCAB_SIZE)
```

```python

def case_text_to_string(text):
    if isinstance(text, str):
        return text

    if isinstance(text, list):
        parts = []
        for item in text:
            if isinstance(item, str):
                parts.append(item)
            elif isinstance(item, dict) and "text" in item:
                parts.append(str(item["text"]))
            else:
                raise TypeError(
                    f"Unsupported text item: {type(item)}"
                )
        return " ".join(parts)

    raise TypeError(f"Unsupported text type: {type(text)}")


def prepare_records(raw_records, labeled=False):
    records = {}

    if isinstance(raw_records, dict):
        iterator = raw_records.items()
    elif isinstance(raw_records, list):
        iterator = (
            (
                row.get(
                    "id",
                    row.get("query_id", row.get("case_id")),
                ),
                row,
            )
            for row in raw_records
        )
    else:
        raise TypeError("Dataset must be a dict or list.")

    for raw_id, row in iterator:
        if raw_id is None:
            raise ValueError("A record has no case ID.")

        case_id = str(raw_id)

        if case_id in records:
            raise ValueError(f"Duplicate case ID: {case_id}")

        text = case_text_to_string(row["text"])

        if not text.strip():
            raise ValueError(f"Empty text for case {case_id}")

        item = {"text": text}

        if labeled:
            positives = row["relevant_candidates"]
            if not isinstance(positives, list):
                raise ValueError(
                    "relevant_candidates must be a list."
                )
            item["relevant_candidates"] = [
                str(cid) for cid in positives
            ]

        records[case_id] = item

    return records



train_data = prepare_records(
    train_data_raw,
    labeled=True,
)
train_candidates = prepare_records(
    train_candidates_raw,
)

dev_data = prepare_records(
    dev_data_raw,
    labeled=True,
)
dev_candidates = prepare_records(
    dev_candidates_raw,
)

test_data = prepare_records(test_data_raw)
test_candidates = prepare_records(test_candidates_raw)

print("Train:", len(train_data), "queries,", len(train_candidates), "candidates")
print("Dev:  ", len(dev_data), "queries,", len(dev_candidates), "candidates")
print("Test: ", len(test_data), "queries,", len(test_candidates), "candidates")
```

```python

def chunk_token_ids(token_ids):
    content = list(token_ids)

    if not content:
        content = [UNK_ID]

    starts = list(range(
        0,
        max(len(content) - CHUNK_CONTENT_LEN, 0) + 1,
        CHUNK_STRIDE,
    ))

    last_start = max(len(content) - CHUNK_CONTENT_LEN, 0)
    if not starts or starts[-1] != last_start:
        starts.append(last_start)

    starts = sorted(set(starts))

    if len(starts) > MAX_CHUNKS:
       
        idx = np.linspace(0, len(starts) - 1, MAX_CHUNKS).round().astype(int)
        starts = [starts[i] for i in idx]
        starts = sorted(set(starts))

    chunks = []
    for start in starts:
        piece = content[start:start + CHUNK_CONTENT_LEN]
        chunk = [BOS_ID] + piece + [EOS_ID]
        chunk = chunk[:MAX_SEQ_LEN]

        if len(chunk) < MAX_SEQ_LEN:
            chunk = chunk + [PAD_ID] * (MAX_SEQ_LEN - len(chunk))

        chunks.append(chunk)

    while len(chunks) < MAX_CHUNKS:
        chunks.append([PAD_ID] * MAX_SEQ_LEN)

    return torch.tensor(chunks[:MAX_CHUNKS], dtype=torch.long)

def encode_record_text(text):
    ids = convert_text_to_ids(text)
    return chunk_token_ids(ids)

def build_chunk_records(records, desc):
    result = {}
    for doc_id, row in tqdm(records.items(), desc=desc):
        result[doc_id] = {
            "chunks": encode_record_text(row["text"])
        }
    return result

train_query_records = build_chunk_records(train_data, "Chunking train queries")
train_candidate_records = build_chunk_records(
    train_candidates, "Chunking train candidates"
)
dev_query_records = build_chunk_records(dev_data, "Chunking dev queries")
dev_candidate_records = build_chunk_records(
    dev_candidates, "Chunking dev candidates"
)
test_query_records = build_chunk_records(test_data, "Chunking test queries")
test_candidate_records = build_chunk_records(
    test_candidates, "Chunking test candidates"
)

print("Example query chunks:", next(iter(train_query_records.values()))["chunks"].shape)
```

```python
# -------------------------
# PCR Dataset
# -------------------------
class PCRDataset(Dataset):
    def __init__(
        self,
        query_data,
        query_records,
        candidate_records,
        negative_map,
        num_negatives=4,
    ):
        self.query_data = query_data
        self.query_records = query_records
        self.candidate_records = candidate_records
        self.negative_map = negative_map
        self.num_negatives = num_negatives

        self.query_ids = [
            qid for qid, row in query_data.items()
            if row.get("relevant_candidates")
        ]

    def __len__(self):
        return len(self.query_ids)

    def __getitem__(self, idx):
        qid = self.query_ids[idx]
        row = self.query_data[qid]

        positives = list(dict.fromkeys(row["relevant_candidates"]))
        negatives = self.negative_map[qid][:self.num_negatives]

        candidate_ids = positives + negatives
        positive_mask = [True] * len(positives) + [False] * len(negatives)

        query_chunks = self.query_records[qid]["chunks"]

        candidate_chunks = torch.stack([
            self.candidate_records[cid]["chunks"]
            for cid in candidate_ids
        ])

        return {
            "qid": qid,
            "query_chunks": query_chunks,
            "candidate_chunks": candidate_chunks,
            "candidate_ids": candidate_ids,
            "positive_mask": torch.tensor(
                positive_mask, dtype=torch.bool
            ),
        }

train_dataset = PCRDataset(
    train_data,
    train_query_records,
    train_candidate_records,
    random_negative_map,
    NUM_NEGATIVES,
)

def collate_fn(batch):

    return batch

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    collate_fn=collate_fn,
)

print("Training queries:", len(train_dataset))
```

```python
# ============================================================
# Supplied GQA Linear Attention kernel
# ============================================================

def chunked_gqa_linear_attn(
    q,
    k,
    v,
    chunk=128,
    eps=1e-6,
    padding_mask=None,
):
    """
    q: (B, Hq, T, F)
    k: (B, Hkv, T, F)
    v: (B, Hkv, T, D)

    All attention-kernel calculations are performed in float32.

    IMPORTANT:
        padding_mask=True means PAD.
        This matches the supplied kernel implementation.
    """
    B, Hq, T, Fd = q.shape

    Hkv = k.shape[1]
    D = v.shape[-1]

    R = Hq // Hkv

    if padding_mask is not None:
        valid = (
            ~padding_mask.bool()
        )[:, None, :, None]

        k = k * valid.to(k.dtype)
        v = v * valid.to(v.dtype)

    q = q.view(
        B,
        Hkv,
        R,
        T,
        Fd,
    )

    S = q.new_zeros(
        B,
        Hkv,
        Fd,
        D,
    )

    z = q.new_zeros(
        B,
        Hkv,
        Fd,
    )

    tri = torch.tril(
        torch.ones(
            chunk,
            chunk,
            dtype=torch.bool,
            device=q.device,
        )
    )

    outs = []

    for s in range(
        0,
        T,
        chunk,
    ):
        qc = q[
            :,
            :,
            :,
            s:s + chunk,
        ]

        kc = k[
            :,
            :,
            s:s + chunk,
        ]

        vc = v[
            :,
            :,
            s:s + chunk,
        ]

        c = qc.shape[3]

        a = (
            qc
            @ kc.unsqueeze(2).transpose(
                -1,
                -2,
            )
        ).masked_fill(
            ~tri[:c, :c],
            0.0,
        )

        num = (
            qc
            @ S.unsqueeze(2)
            + a @ vc.unsqueeze(2)
        )

        den = (
            (
                qc
                * z[
                    :,
                    :,
                    None,
                    None,
                    :,
                ]
            ).sum(
                -1,
                keepdim=True,
            )
            + a.sum(
                -1,
                keepdim=True,
            )
        )

        outs.append(
            num / (den + eps)
        )

        S = (
            S
            + kc.transpose(
                -1,
                -2,
            ) @ vc
        )

        z = (
            z
            + kc.sum(dim=2)
        )

    return (
        torch.cat(
            outs,
            dim=3,
        ).reshape(
            B,
            Hq,
            T,
            D,
        ),
        (S, z),
    )


class LinearAttention(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        dropout=0.0,
        feat_dim=32,
        chunk=128,
    ):
        super().__init__()

        assert (
            d_model % n_heads == 0
            and n_heads % n_kv_heads == 0
        )

        self.n_heads = n_heads
        self.n_kv_heads = n_kv_heads
        self.n_rep = n_heads // n_kv_heads

        self.d_head = (
            d_model // n_heads
        )

        self.chunk = chunk
        self.feat_dim = feat_dim

        self.q_proj = nn.Linear(
            d_model,
            n_heads * self.d_head,
            bias=False,
        )

        self.k_proj = nn.Linear(
            d_model,
            n_kv_heads * self.d_head,
            bias=False,
        )

        self.v_proj = nn.Linear(
            d_model,
            n_kv_heads * self.d_head,
            bias=False,
        )

        self.out_proj = nn.Linear(
            n_heads * self.d_head,
            d_model,
            bias=False,
        )

        # phi(x) = (LayerNorm(gamma*x + beta))^2
        self.q_kernel = nn.Linear(
            self.d_head,
            feat_dim,
            bias=True,
        )

        self.k_kernel = nn.Linear(
            self.d_head,
            feat_dim,
            bias=True,
        )

        self.q_norm = nn.LayerNorm(
            feat_dim
        )

        self.k_norm = nn.LayerNorm(
            feat_dim
        )

        self.resid_dropout = nn.Dropout(
            dropout
        )

    def _project(self, x):
        B, T, _ = x.shape

        q = self.q_proj(x).view(
            B,
            T,
            self.n_heads,
            self.d_head,
        ).transpose(1, 2)

        k = self.k_proj(x).view(
            B,
            T,
            self.n_kv_heads,
            self.d_head,
        ).transpose(1, 2)

        v = self.v_proj(x).view(
            B,
            T,
            self.n_kv_heads,
            self.d_head,
        ).transpose(1, 2)

        q_phi = (
            self.q_norm(
                self.q_kernel(q)
            ).float()
            ** 2
        )

        k_phi = (
            self.k_norm(
                self.k_kernel(k)
            ).float()
            ** 2
        )

        return (
            q_phi,
            k_phi,
            v.float(),
        )

    def forward(
        self,
        x,
        padding_mask=None,
        return_state=False,
    ):
        B, T, C = x.shape

        q_phi, k_phi, v = (
            self._project(x)
        )

        # The supplied kernel is explicitly float32.
        with torch.autocast(
            device_type="cuda",
            enabled=False,
        ):
            out, state = (
                chunked_gqa_linear_attn(
                    q_phi,
                    k_phi,
                    v,
                    self.chunk,
                    padding_mask=padding_mask,
                )
            )

        out = F.rms_norm(
            out.to(x.dtype),
            (self.d_head,),
        )

        out = self.out_proj(
            out.transpose(
                1,
                2,
            ).reshape(
                B,
                T,
                C,
            )
        )

        out = self.resid_dropout(
            out
        )

        return (
            (out, state)
            if return_state
            else out
        )

    @torch.no_grad()
    def step(
        self,
        x_t,
        S,
        z,
        eps=1e-6,
    ):
        """
        One-token decoding.

        x_t: (B,1,C)
        S:   (B,Hkv,F,D), float32
        z:   (B,Hkv,F), float32
        """
        B = x_t.shape[0]

        q_phi, k_phi, v = (
            self._project(x_t)
        )

        S = (
            S
            + k_phi.transpose(
                -1,
                -2,
            ) @ v
        )

        z = (
            z
            + k_phi.squeeze(2)
        )

        qg = q_phi.reshape(
            B,
            self.n_kv_heads,
            self.n_rep,
            self.feat_dim,
        )

        numerator = torch.einsum(
            "bhrf,bhfd->bhrd",
            qg,
            S,
        )

        denominator = (
            qg
            * z.unsqueeze(2)
        ).sum(
            -1,
            keepdim=True,
        )

        out = numerator / (
            denominator + eps
        )

        out = F.rms_norm(
            out.to(x_t.dtype),
            (self.d_head,),
        )

        out = out.reshape(
            B,
            1,
            self.n_heads
            * self.d_head,
        )

        out = self.out_proj(
            out
        )

        return (
            out,
            S,
            z,
        )
```

```python
# -------------------------
# Positional encoding
# -------------------------
# LinearAttention itself does not apply RoPE.
# PCR uses the supplied decoder's sinusoidal positional encoding.
```

```python
# ============================================================
# Decoder-only Transformer LM with LinearAttention
# ============================================================

@dataclass
class LMConfig:
    vocab_size: int
    d_model: int = 512
    n_layers: int = 12
    n_heads: int = 8
    n_kv_heads: int = 8
    d_ff: int = 4096
    max_seq_len: int = 1024
    dropout: float = 0.1
    pos_emb_type: str = "sinusoidal"
    feat_dim: int = 32
    chunk: int = 128


def sinusoidal_positional_encoding(
    max_len,
    d_model,
):
    pos = torch.arange(
        max_len,
        dtype=torch.float32,
    )[:, None]

    i = torch.arange(
        d_model // 2,
        dtype=torch.float32,
    )[None, :]

    ang = pos / (
        10000
        ** (
            2 * i / d_model
        )
    )

    pe = torch.zeros(
        max_len,
        d_model,
    )

    pe[:, 0::2] = torch.sin(
        ang
    )
    pe[:, 1::2] = torch.cos(
        ang
    )

    return pe


class FeedForward(nn.Module):
    def __init__(
        self,
        d_model,
        d_ff,
        dropout,
    ):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(
                d_model,
                d_ff,
            ),
            nn.GELU(),
            nn.Linear(
                d_ff,
                d_model,
            ),
            nn.Dropout(
                dropout
            ),
        )

    def forward(self, x):
        return self.net(x)


class Block(nn.Module):
    def __init__(
        self,
        cfg,
    ):
        super().__init__()

        self.ln1 = nn.LayerNorm(
            cfg.d_model
        )

        self.attn = LinearAttention(
            cfg.d_model,
            cfg.n_heads,
            cfg.n_kv_heads,
            cfg.dropout,
            cfg.feat_dim,
            cfg.chunk,
        )

        self.ln2 = nn.LayerNorm(
            cfg.d_model
        )

        self.ffn = FeedForward(
            cfg.d_model,
            cfg.d_ff,
            cfg.dropout,
        )

    def forward(
        self,
        x,
        padding_mask=None,
        return_state=False,
    ):
        if return_state:
            a, state = self.attn(
                self.ln1(x),
                padding_mask=padding_mask,
                return_state=True,
            )
        else:
            a = self.attn(
                self.ln1(x),
                padding_mask=padding_mask,
            )
            state = None

        x = x + a

        x = x + self.ffn(
            self.ln2(x)
        )

        return (
            (x, state)
            if return_state
            else x
        )

    @torch.no_grad()
    def step(
        self,
        x_t,
        state,
    ):
        S, z = state

        a, S, z = (
            self.attn.step(
                self.ln1(x_t),
                S,
                z,
            )
        )

        x = x_t + a

        x = x + self.ffn(
            self.ln2(x)
        )

        return x, (S, z)


class DecoderLM(nn.Module):
    def __init__(
        self,
        cfg,
    ):
        super().__init__()

        self.cfg = cfg

        self.tok_emb = nn.Embedding(
            cfg.vocab_size,
            cfg.d_model,
        )

        if (
            cfg.pos_emb_type
            == "sinusoidal"
        ):
            self.register_buffer(
                "pos_emb",
                sinusoidal_positional_encoding(
                    cfg.max_seq_len,
                    cfg.d_model,
                ),
                persistent=False,
            )

        self.blocks = nn.ModuleList(
            [
                Block(cfg)
                for _ in range(
                    cfg.n_layers
                )
            ]
        )

        self.ln_f = nn.LayerNorm(
            cfg.d_model
        )

        self.head = nn.Linear(
            cfg.d_model,
            cfg.vocab_size,
            bias=False,
        )

        self.head.weight = (
            self.tok_emb.weight
        )

        with torch.no_grad():
            nn.init.uniform_(
                self.tok_emb.weight,
                -1
                / math.sqrt(
                    cfg.d_model
                ),
                1
                / math.sqrt(
                    cfg.d_model
                ),
            )

    def encode(
        self,
        idx,
        padding_mask=None,
    ):
        x = self.tok_emb(idx)

        if (
            self.cfg.pos_emb_type
            == "sinusoidal"
        ):
            T = idx.shape[1]

            x = x + self.pos_emb[
                :T
            ].unsqueeze(0)

        for block in self.blocks:
            if (
                self.training
                and USE_GRAD_CHECKPOINTING
                and torch.is_grad_enabled()
            ):
                x = torch.utils.checkpoint.checkpoint(
                    block,
                    x,
                    padding_mask,
                    use_reentrant=False,
                )
            else:
                x = block(
                    x,
                    padding_mask=padding_mask,
                )

        return self.ln_f(x)

    def forward(
        self,
        idx,
        padding_mask=None,
    ):
        x = self.encode(
            idx,
            padding_mask=padding_mask,
        )

        return self.head(x)

    @torch.no_grad()
    def prefill(self, idx):
        x = self.tok_emb(idx)

        if (
            self.cfg.pos_emb_type
            == "sinusoidal"
        ):
            T = idx.shape[1]
            x = x + self.pos_emb[
                :T
            ].unsqueeze(0)

        states = []

        for block in self.blocks:
            x, state = block(
                x,
                return_state=True,
            )
            states.append(state)

        logits = self.head(
            self.ln_f(x[:, -1:])
        )

        return logits, states

    @torch.no_grad()
    def step(
        self,
        tok,
        pos,
        states,
    ):
        x = self.tok_emb(tok)

        if (
            self.cfg.pos_emb_type
            == "sinusoidal"
        ):
            x = x + self.pos_emb[
                pos:pos + 1
            ].unsqueeze(0)

        new_states = []

        for block, state in zip(
            self.blocks,
            states,
        ):
            x, new_state = (
                block.step(
                    x,
                    state,
                )
            )
            new_states.append(
                new_state
            )

        logits = self.head(
            self.ln_f(x)
        )

        return logits, new_states
```

```python
# -------------------------
# Standard additive LoRA
# -------------------------
class LoRALinear(nn.Module):
    """
    Standard additive LoRA:

        y = W0 x + (alpha/r) B A x

    W0 is frozen.
    A and B are trainable.
    """
    def __init__(
        self,
        layer,
        rank=16,
        alpha=32,
        dropout=0.05,
    ):
        super().__init__()

        if not isinstance(layer, nn.Linear):
            raise TypeError("LoRALinear expects nn.Linear")

        self.layer = layer
        self.rank = int(rank)
        self.alpha = float(alpha)
        self.scaling = self.alpha / self.rank
        self.dropout = nn.Dropout(dropout)

        for p in self.layer.parameters():
            p.requires_grad_(False)

        self.A = nn.Parameter(
            torch.empty(
                self.rank,
                layer.in_features,
            )
        )
        self.B = nn.Parameter(
            torch.zeros(
                layer.out_features,
                self.rank,
            )
        )

        nn.init.kaiming_uniform_(
            self.A,
            a=5**0.5,
        )

    def forward(self, x):
        base_output = self.layer(x)

        low_rank = F.linear(
            self.dropout(x),
            self.A,
        )

        update = F.linear(
            low_rank,
            self.B,
        )

        return base_output + self.scaling * update
```

```python
# ============================================================
# Build LinearAttention model and inject standard LoRA
# ============================================================

cfg = LMConfig(
    vocab_size=VOCAB_SIZE,
    d_model=512,
    n_layers=12,
    n_heads=8,
    n_kv_heads=8,
    d_ff=4096,
    max_seq_len=MAX_SEQ_LEN,
    dropout=0.1,
    pos_emb_type="sinusoidal",
    feat_dim=32,
    chunk=128,
)

model = DecoderLM(cfg)
BASE_CHECKPOINT_PATH = "/kaggle/input/datasets/rogue2222/kernel-mha/kernel_final.pt"
if not os.path.exists(
    BASE_CHECKPOINT_PATH
):
    raise FileNotFoundError(
        f"Base decoder checkpoint not found: "
        f"{BASE_CHECKPOINT_PATH}\n"
        "Set BASE_CHECKPOINT_PATH to a checkpoint "
        "compatible with this LinearAttention architecture."
    )

checkpoint = torch.load(
    BASE_CHECKPOINT_PATH,
    map_location="cpu",
    weights_only=False,
)

if (
    isinstance(checkpoint, dict)
    and "model_state_dict" in checkpoint
):
    state_dict = checkpoint[
        "model_state_dict"
    ]
else:
    state_dict = checkpoint

state_dict = {
    (
        k.replace(
            "module.",
            "",
            1,
        )
        if k.startswith("module.")
        else k
    ): v
    for k, v in state_dict.items()
}


current_state = model.state_dict()

compatible_state = {}
skipped_state = {}

for key, value in state_dict.items():
    if key not in current_state:
        skipped_state[key] = "not in LinearAttention model"
        continue

    if current_state[key].shape != value.shape:
        skipped_state[key] = (
            f"shape mismatch: checkpoint={tuple(value.shape)}, "
            f"model={tuple(current_state[key].shape)}"
        )
        continue

    compatible_state[key] = value

missing, unexpected = model.load_state_dict(
    compatible_state,
    strict=False,
)

print(
    "Compatible checkpoint tensors loaded:",
    len(compatible_state),
)
print(
    "Missing keys:",
    len(missing),
)
print(
    "Skipped checkpoint keys:",
    len(skipped_state),
)

print(
    "\nMissing keys before LoRA injection:"
)
for key in missing:
    print("  ", key)

if skipped_state:
    print(
        "\nSkipped checkpoint keys:"
    )
    for key, reason in list(
        skipped_state.items()
    )[:20]:
        print(
            "  ",
            key,
            "->",
            reason,
        )

# ------------------------------------------------------------
# Inject standard additive LoRA into the four projections.
# ------------------------------------------------------------
target_names = (
    "q_proj",
    "k_proj",
    "v_proj",
    "out_proj",
)

for block in model.blocks:
    for name in target_names:
        old_layer = getattr(
            block.attn,
            name,
        )

        setattr(
            block.attn,
            name,
            LoRALinear(
                old_layer,
                rank=LORA_RANK,
                alpha=LORA_ALPHA,
                dropout=LORA_DROPOUT,
            ),
        )

model = model.to(device)

# ------------------------------------------------------------
# Freeze everything except LoRA A/B.
# ------------------------------------------------------------
for name, p in model.named_parameters():
    if (
        ".A" in name
        or ".B" in name
    ):
        p.requires_grad_(True)
    else:
        p.requires_grad_(False)

trainable = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

total = sum(
    p.numel()
    for p in model.parameters()
)

print(
    f"\nTrainable: {trainable:,}"
)
print(
    f"Total:     {total:,}"
)
print(
    f"Trainable %: "
    f"{100 * trainable / total:.4f}%"
)
```

```python
# ============================================================
# Encode documents
# ============================================================

def mean_pool_tokens(
    hidden,
    valid_mask,
):
    """
    valid_mask=True means VALID token.
    """
    mask = valid_mask.unsqueeze(
        -1
    ).to(
        hidden.dtype
    )

    return (
        (hidden * mask).sum(
            dim=1
        )
        / mask.sum(
            dim=1
        ).clamp_min(1.0)
    )


def encode_query_tensor(
    model,
    query_chunks,
):
    """
    Encode one document.

    Important mask conventions:
      * token_valid_mask=True  -> VALID
      * attention padding_mask=True -> PAD
        (required by chunked_gqa_linear_attn)
    """
    chunks = query_chunks.to(
        device,
        non_blocking=True,
    )

    valid = (
        chunks != PAD_ID
    ).any(dim=1)

    valid_chunks = chunks[
        valid
    ]

    if valid_chunks.size(0) == 0:
        return F.normalize(
            torch.zeros(
                model.cfg.d_model,
                device=device,
            ),
            p=2,
            dim=-1,
        )

    chunk_embs = []

    for j in range(
        0,
        valid_chunks.size(0),
        CHUNK_BATCH_SIZE,
    ):
        token_ids = valid_chunks[
            j:j + CHUNK_BATCH_SIZE
        ]

        # Pooling needs TRUE for valid tokens.
        token_valid_mask = (
            token_ids != PAD_ID
        )

        # Supplied LinearAttention kernel needs TRUE for PAD.
        attention_padding_mask = (
            token_ids == PAD_ID
        )

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=(
                USE_AMP
                and torch.cuda.is_available()
            ),
        ):
            hidden = model.encode(
                token_ids,
                padding_mask=attention_padding_mask,
            )

            pooled = mean_pool_tokens(
                hidden,
                token_valid_mask,
            )

        chunk_embs.append(
            pooled.float()
        )

        del (
            token_ids,
            token_valid_mask,
            attention_padding_mask,
            hidden,
            pooled,
        )

    chunk_embs = torch.cat(
        chunk_embs,
        dim=0,
    )

    document_emb = chunk_embs.mean(
        dim=0
    )

    return F.normalize(
        document_emb.float(),
        p=2,
        dim=-1,
    )


@torch.no_grad()
def encode_documents(
    model,
    records,
    document_ids,
    device,
    chunk_batch_size=1,
):
    model.eval()

    document_embeddings = []

    for doc_start in tqdm(
        range(
            len(document_ids)
        ),
        desc="Encoding documents",
    ):
        doc_id = document_ids[
            doc_start
        ]

        chunks = records[
            doc_id
        ]["chunks"]

        emb = encode_query_tensor(
            model,
            chunks,
        )

        document_embeddings.append(
            emb.cpu()
        )

        del chunks, emb

        if (
            torch.cuda.is_available()
            and doc_start % 25 == 0
        ):
            torch.cuda.empty_cache()

    return torch.stack(
        document_embeddings,
        dim=0,
    )
```

```python
# -------------------------
# Multi-positive contrastive loss
# -------------------------
def multi_positive_contrastive_loss(
    similarity_scores,
    positive_mask,
    candidate_mask,
    temperature=0.07,
):
    logits = similarity_scores / temperature

    logits = logits.masked_fill(
        ~candidate_mask,
        float("-inf"),
    )

    positive_logits = logits.masked_fill(
        ~positive_mask,
        float("-inf"),
    )

    valid_rows = (
        positive_mask.any(dim=1)
        & candidate_mask.any(dim=1)
    )

    if not valid_rows.all():
        raise ValueError(
            "Every query must have at least one positive "
            "and one candidate."
        )

    numerator = torch.logsumexp(
        positive_logits,
        dim=1,
    )

    denominator = torch.logsumexp(
        logits,
        dim=1,
    )

    return -(numerator - denominator).mean()
```

```python
# -------------------------
# One training epoch
# Memory-efficient LoRA PCR training
# -------------------------
def train_one_epoch(
    model,
    train_loader,
    optimizer,
    scaler,
    device,
):
    model.train()

    running_loss = 0.0
    count = 0

    for batch in tqdm(
        train_loader,
        desc="Training",
        leave=False,
    ):
        # collate_fn returns a list for BATCH_SIZE > 1.
        samples = (
            batch
            if isinstance(batch, list)
            else [batch]
        )

        batch_size_actual = len(
            samples
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        batch_loss = 0.0

        # -------------------------------------------------
        # IMPORTANT:
        # Each query is fully processed and BACKWARDED
        # before moving to the next query.
        #
        # Therefore BATCH_SIZE=2 does NOT keep two complete
        # Transformer activation graphs in memory.
        # -------------------------------------------------
        for sample_idx, sample in enumerate(
            samples
        ):
            query_chunks = sample[
                "query_chunks"
            ].to(
                device,
                non_blocking=True,
            )

            candidate_chunks = sample[
                "candidate_chunks"
            ].to(
                device,
                non_blocking=True,
            )

            positive_mask = sample[
                "positive_mask"
            ].to(
                device,
                non_blocking=True,
            )

            # ---------------------------------------------
            # Query encoding
            # ---------------------------------------------
            query_embedding = (
                encode_query_tensor(
                    model,
                    query_chunks,
                )
            )

            # ---------------------------------------------
            # Candidate encoding
            # ---------------------------------------------
            candidate_embeddings = []

            for c in range(
                candidate_chunks.size(0)
            ):
                candidate_embedding = (
                    encode_query_tensor(
                        model,
                        candidate_chunks[c],
                    )
                )

                candidate_embeddings.append(
                    candidate_embedding
                )

            candidate_embeddings = torch.stack(
                candidate_embeddings,
                dim=0,
            )

            similarities = (
                query_embedding.unsqueeze(0)
                @ candidate_embeddings.transpose(
                    0,
                    1,
                )
            )

            loss = (
                multi_positive_contrastive_loss(
                    similarities,
                    positive_mask.unsqueeze(0),
                    torch.ones_like(
                        positive_mask,
                        dtype=torch.bool,
                    ).unsqueeze(0),
                    temperature=TEMPERATURE,
                )
            )

            # Average the gradients over queries in this batch.
            loss_for_backward = (
                loss / batch_size_actual
            )

            if USE_AMP and torch.cuda.is_available():
                scaler.scale(
                    loss_for_backward
                ).backward()
            else:
                loss_for_backward.backward()

            batch_loss += float(
                loss.detach().item()
            )

            # Free everything belonging to this query.
            del (
                query_chunks,
                candidate_chunks,
                positive_mask,
                query_embedding,
                candidate_embeddings,
                similarities,
                loss,
                loss_for_backward,
            )

            # The autograd graph from the query is released after backward.
            # This keeps the peak memory approximately independent of BATCH_SIZE.

            if torch.cuda.is_available():
                torch.cuda.empty_cache()

        # ---------------------------------------------
        # One optimizer update for the whole batch
        # ---------------------------------------------
        if USE_AMP and torch.cuda.is_available():
            scaler.unscale_(optimizer)

        torch.nn.utils.clip_grad_norm_(
            [
                p
                for p in model.parameters()
                if p.requires_grad
            ],
            GRAD_CLIP,
        )

        if USE_AMP and torch.cuda.is_available():
            scaler.step(optimizer)
            scaler.update()
        else:
            optimizer.step()

        batch_loss /= batch_size_actual

        running_loss += batch_loss
        count += 1

        del samples

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    return (
        running_loss
        / max(count, 1)
    )
```

```python
# -------------------------
# Full-pool retrieval evaluation
# -------------------------
@torch.no_grad()
def evaluate_retrieval_full_pool(
    model,
    query_records,
    candidate_records,
    query_data,
    k=7,
    eval_batch_size=1,
):
    model.eval()

    candidate_ids = list(
        candidate_records.keys()
    )

    query_ids = [
        qid
        for qid in query_data.keys()
        if query_data[qid].get(
            "relevant_candidates"
        )
    ]

    candidate_embeddings = (
        encode_documents(
            model,
            candidate_records,
            candidate_ids,
            device,
            chunk_batch_size=1,
        )
    )

    query_embeddings = []

    for qid in tqdm(
        query_ids,
        desc="Encoding queries",
    ):
        emb = encode_query_tensor(
            model,
            query_records[qid]["chunks"],
        )
        query_embeddings.append(
            emb.cpu()
        )

    query_embeddings = torch.stack(
        query_embeddings
    )

    hits = 0
    total_relevant = 0
    total_retrieved = (
        len(query_ids) * k
    )

    predictions = {}

    for i, qid in enumerate(
        query_ids
    ):
        scores = (
            query_embeddings[
                i:i + 1
            ]
            @ candidate_embeddings.T
        ).squeeze(0)

        topk = min(
            k,
            scores.numel(),
        )

        values, indices = torch.topk(
            scores,
            k=topk,
        )

        retrieved = [
            candidate_ids[int(j)]
            for j in indices
        ]

        relevant = set(
            query_data[qid][
                "relevant_candidates"
            ]
        )

        intersection = len(
            set(retrieved)
            & relevant
        )

        hits += intersection
        total_relevant += len(
            relevant
        )

        predictions[qid] = {
            "retrieved_candidates": retrieved,
            "relevant_candidates": list(
                relevant
            ),
            "scores": [
                float(v)
                for v in values
            ],
        }

    precision = (
        hits / total_retrieved
        if total_retrieved
        else 0.0
    )

    recall = (
        hits / total_relevant
        if total_relevant
        else 0.0
    )

    f1 = (
        2 * precision * recall
        / (precision + recall)
        if precision + recall > 0
        else 0.0
    )

    return {
        "Micro-Precision@7": precision,
        "Micro-Recall@7": recall,
        "Micro-F1@7": f1,
    }, predictions
```

```python
# -------------------------
# Checkpoint helper
# -------------------------
def save_checkpoint(
    path,
    epoch,
    model,
    optimizer,
    scaler,
    train_loss,
    metrics=None,
):
    payload = {
        "epoch": epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scaler_state_dict": scaler.state_dict(),
        "train_loss": train_loss,
        "metrics": metrics,
        "config": {
            "max_seq_len": MAX_SEQ_LEN,
            "chunk_overlap": CHUNK_OVERLAP,
            "max_chunks": MAX_CHUNKS,
            "num_negatives": NUM_NEGATIVES,
            "chunk_batch_size": CHUNK_BATCH_SIZE,
            "temperature": TEMPERATURE,
            "lora_rank": LORA_RANK,
            "lora_alpha": LORA_ALPHA,
            "lora_dropout": LORA_DROPOUT,
            "lora_a_lr": LORA_A_LR,
            "lora_b_lr": LORA_B_LR,
            "adaptation": "standard additive LoRA + random negatives",
            "backbone": "decoder-only GQA LinearAttention",
        },
    }

    torch.save(
        payload,
        path,
    )

    return path
```

```python
# -------------------------
# Train
# -------------------------
best_f1 = -1.0
history = []

for epoch in range(
    1,
    EPOCHS + 1,
):
    print(
        f"\n===== Epoch {epoch}/{EPOCHS} ====="
    )

    clear_cuda_memory()
    gpu_memory_report(
        "Before train | "
    )

    train_loss = train_one_epoch(
        model=model,
        train_loader=train_loader,
        optimizer=optimizer,
        scaler=scaler,
        device=device,
    )

    gpu_memory_report(
        "After train | "
    )

    epoch_checkpoint = os.path.join(
        CHECKPOINT_DIR,
        f"epoch_{epoch}.pt",
    )

    save_checkpoint(
        epoch_checkpoint,
        epoch,
        model,
        optimizer,
        scaler,
        train_loss,
        metrics=None,
    )

    dev_metrics, dev_predictions = (
        evaluate_retrieval_full_pool(
            model=model,
            query_records=dev_query_records,
            candidate_records=dev_candidate_records,
            query_data=dev_data,
            k=7,
            eval_batch_size=1,
        )
    )

    row = {
        "epoch": epoch,
        "train_loss": train_loss,
        **dev_metrics,
    }

    history.append(row)

    with open(
        HISTORY_PATH,
        "w",
        encoding="utf-8",
    ) as f:
        json.dump(
            history,
            f,
            indent=2,
        )

    print(
        f"Train loss: {train_loss:.6f}"
    )
    print(
        f"Precision@7: "
        f"{dev_metrics['Micro-Precision@7']:.6f}"
    )
    print(
        f"Recall@7: "
        f"{dev_metrics['Micro-Recall@7']:.6f}"
    )
    print(
        f"F1@7: "
        f"{dev_metrics['Micro-F1@7']:.6f}"
    )

    current_f1 = dev_metrics[
        "Micro-F1@7"
    ]

    if current_f1 > best_f1:
        best_f1 = current_f1

        save_checkpoint(
            BEST_CHECKPOINT,
            epoch,
            model,
            optimizer,
            scaler,
            train_loss,
            metrics=dev_metrics,
        )

        with open(
            os.path.join(
                CHECKPOINT_DIR,
                "best_dev_predictions.json",
            ),
            "w",
            encoding="utf-8",
        ) as f:
            json.dump(
                dev_predictions,
                f,
                indent=2,
            )

        print(
            "Saved new best checkpoint:",
            BEST_CHECKPOINT,
        )

    clear_cuda_memory()
```

```python


def load_lora_checkpoint(
    model,
    checkpoint_path,
):
    if not os.path.exists(
        checkpoint_path
    ):
        raise FileNotFoundError(
            f"Checkpoint not found: "
            f"{checkpoint_path}"
        )

    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False,
    )

    if (
        "model_state_dict"
        in checkpoint
    ):
        state_dict = checkpoint[
            "model_state_dict"
        ]
    elif (
        "lora_state_dict"
        in checkpoint
    ):
        state_dict = checkpoint[
            "lora_state_dict"
        ]
    else:
        state_dict = checkpoint

    missing, unexpected = (
        model.load_state_dict(
            state_dict,
            strict=False,
        )
    )

    print(
        "Loaded checkpoint:",
        checkpoint_path,
    )

    if isinstance(
        checkpoint,
        dict,
    ):
        print(
            "Checkpoint epoch:",
            checkpoint.get(
                "epoch"
            ),
        )
        print(
            "Checkpoint metrics:",
            checkpoint.get(
                "metrics"
            ),
        )

    # Strictness is relaxed because a LoRA-only checkpoint
    # intentionally does not contain frozen base parameters.
    if missing:
        print(
            "Missing keys:",
            len(missing),
        )

    if unexpected:
        print(
            "Unexpected keys:",
            len(unexpected),
        )

    return checkpoint


def find_latest_epoch_checkpoint():
    epoch_paths = glob.glob(
        os.path.join(
            CHECKPOINT_DIR,
            "epoch_*.pt",
        )
    )

    if not epoch_paths:
        return None

    def epoch_num(path):
        match = re.search(
            r"epoch_(\d+)\.pt$",
            path,
        )
        return (
            int(match.group(1))
            if match
            else -1
        )

    return max(
        epoch_paths,
        key=epoch_num,
    )


@torch.no_grad()
def generate_test_retrieval(
    model,
    query_records,
    candidate_records,
    test_data,
    top_k=7,
):
    model.eval()

    candidate_ids = list(
        candidate_records.keys()
    )

    print(
        "Encoding test candidates:",
        len(candidate_ids),
    )

    candidate_embeddings = (
        encode_documents(
            model,
            candidate_records,
            candidate_ids,
            device,
            chunk_batch_size=1,
        )
    )

    query_ids = list(
        test_data.keys()
    )

    predictions = {}

    for qid in tqdm(
        query_ids,
        desc="Generating test retrieval",
    ):
        query_embedding = (
            encode_query_tensor(
                model,
                query_records[
                    qid
                ]["chunks"],
            ).cpu()
        )

        scores = (
            query_embedding.unsqueeze(
                0
            )
            @ candidate_embeddings.T
        ).squeeze(0)

        k = min(
            top_k,
            scores.numel(),
        )

        values, indices = (
            torch.topk(
                scores,
                k=k,
            )
        )

        retrieved_candidates = [
            candidate_ids[
                int(idx)
            ]
            for idx in indices
        ]

        predictions[str(qid)] = {
            "retrieved_candidates":
                retrieved_candidates,
            "scores": [
                float(v)
                for v in values
            ],
        }

        del (
            query_embedding,
            scores,
            values,
            indices,
        )

    return predictions


clear_cuda_memory()

if os.path.exists(
    BEST_CHECKPOINT
):
    TEST_CHECKPOINT = (
        BEST_CHECKPOINT
    )
else:
    TEST_CHECKPOINT = (
        find_latest_epoch_checkpoint()
    )

if TEST_CHECKPOINT is None:
    raise FileNotFoundError(
        "\nNo PCR checkpoint exists.\n"
        f"Expected best checkpoint:\n{BEST_CHECKPOINT}\n"
        f"Checkpoint directory:\n{CHECKPOINT_DIR}\n\n"
        "Run the training cell first."
    )

best_checkpoint = (
    load_lora_checkpoint(
        model,
        TEST_CHECKPOINT,
    )
)

model = model.to(device)
model.eval()

test_predictions = (
    generate_test_retrieval(
        model=model,
        query_records=test_query_records,
        candidate_records=test_candidate_records,
        test_data=test_data,
        top_k=7,
    )
)

with open(
    TEST_PREDICTIONS_PATH,
    "w",
    encoding="utf-8",
) as f:
    json.dump(
        test_predictions,
        f,
        indent=2,
    )

print(
    f"\nSaved {len(test_predictions)} "
    "test query predictions."
)

print(
    "Output:",
    TEST_PREDICTIONS_PATH,
)

if test_predictions:
    first_qid = next(
        iter(test_predictions)
    )

    print(
        "\nExample test prediction:"
    )
    print(
        first_qid,
        test_predictions[
            first_qid
        ],
    )
```
