```python

class Node:
    def __init__(self,value, previous, next, index):
        self.value=value
        self.previous=previous
        self.next=next
        self.index=index


class LinkedArray:
    def __init__(self, items):
        self._array=[None for _ in range(len(items))]
        previous=None
        for i in range(len(items)):
            node=Node(items[i], previous, None, i)
            if previous is not None:
                node.previous=previous
                previous.next=node
            self._array[i]=node
            previous=node

    def get_previous_index(self, index):
        if self._array[index]==None:
            raise ValueError(f"No item at index {index}")
        if self._array[index].previous == None:
            return None
        return self._array[index].previous.index

    def get_next_index(self, index):
        if self._array[index]==None:
            raise ValueError(f"No item at index {index}")
        if self._array[index].next == None:
            return None
        return self._array[index].next.index

    def get_by_index(self, index):
        if self._array[index]==None:
            raise ValueError(f"No item at index {index}")
        if self._array[index] == None:
            return None
        return self._array[index].value

    def get_second_next_index(self, index):
        if self._array[index] == None:
            raise ValueError(f"No item at index {index}")
        if self._array[index].next == None or self._array[index].next.next == None:
            return None
        return self._array[index].next.next.index

    def len(self):
        return len(self._array)
    def exists(self, index):
        return (
            0 <= index < len(self._array)
            and self._array[index] is not None
        )

    def replace_pair(self, index, new_item):
        if index > len(self._array)-2 or self._array[index] == None or self._array[index].next == None:
            raise ValueError("Invalid index ",index,new_item)
        self._array[index].value = new_item
        self._array[self._array[index].next.index] = None
        self._array[index].next = self._array[index].next.next
        if self._array[index].next is not None:
            self._array[index].next.previous = self._array[index]


class MaxPriorityMap:
    def __init__(self, heap_key, map_key):
        self._heap_key = heap_key
        self._map_key = map_key
        self._heap = []
        # Map of keys to indexes in the heap
        self._map = {}

    def get_max(self):
        if len(self._heap) == 0:
            raise ValueError("the heap is empty")
        return self._heap[0]

    def push(self, item):
        self._heap.append(item)
        map_key = self._map_key(item)
        if map_key in self._map:
            raise ValueError("Map key already exist")
        self._map[self._map_key(item)] = len(self._heap) - 1
        self._heapify_up(len(self._heap) - 1)

    def pop(self):
        if len(self._heap) == 0:
            raise ValueError("the heap is empty")
        self._swap(0, len(self._heap) - 1)
        item = self._heap.pop()
        map_key = self._map_key(item)
        self._map.pop(map_key)
        self._heapify_down(0)
        return item

    def contains(self, map_key):
        return map_key in self._map

    def len(self):
        if len(self._heap) != len(self._map):
            raise ValueError('Invalid state of the priority map')
        return len(self._heap)

    def delete_by_map_key(self, map_key):
        if map_key not in self._map:
            raise ValueError("Map key does not exist")
        heap_index = self._map[map_key]
        item = self._heap[heap_index]
        self._swap(heap_index, len(self._heap) - 1)
        self._heap.pop()
        self._map.pop(map_key)
        if (heap_index < len(self._heap)):
            self._heapify_down(heap_index)
            self._heapify_up(heap_index)
        return item

    def _heapify_up(self, i):
        while i > 0 and self._heap_key(self._heap[i]) > self._heap_key(self._heap[self._parent(i)]):
            self._swap(i, self._parent(i))
            i = self._parent(i)

    def _heapify_down(self, i):
        largest = i
        left = self._left_child(i)
        right = self._right_child(i)

        if left < len(self._heap) and self._heap_key(self._heap[left]) > self._heap_key(self._heap[largest]):
            largest = left

        if right < len(self._heap) and self._heap_key(self._heap[right]) > self._heap_key(self._heap[largest]):
            largest = right

        if largest != i:
            self._swap(i, largest)
            self._heapify_down(largest)

    def _parent(self, i):
        return (i - 1) // 2

    def _left_child(self, i):
        return 2 * i + 1

    def _right_child(self, i):
        return 2 * i + 2

    def _swap(self, i, j):
        self._map[self._map_key(self._heap[i])] = j
        self._map[self._map_key(self._heap[j])] = i
        self._heap[i], self._heap[j] = self._heap[j], self._heap[i]


class StatsEntry:
    def __init__(self, pair, positions):
        self.pair = pair
        self.positions = positions

class TokenizerTrainer:
        def __init__(self, tokens_list, itos, num_merges):
            self.tokens_list = tokens_list
            self.itos = itos
            self._num_merges = num_merges
            self.stoi={}

        def train(self, next_token):
            self._positions=[LinkedArray(tokens) for tokens in self.tokens_list]
            self.get_stats()
            merges_done=0
            print("requested merges",self._num_merges)
            while  merges_done<self._num_merges:
                if self._stats.len() == 0:
                   print("STOPPED: stats became empty")
                   print("Merges completed:", merges_done)
                   break
                merge_stat = self._stats.pop()
                (current_token , next_token) = self.update_itos_and_stoi(next_token, merge_stat)
                #iterate through posistions replace with merge pair
                for position in list(merge_stat.positions):

                    if position not in merge_stat.positions:
                        continue
                    input_index = position[0]
                    token_index = position[1]
                    # print("current_token",current_token)
                    if not self._positions[input_index].exists(token_index):
                        merge_stat.positions.discard(position)
                        continue
                    self.update_right_token(input_index, token_index, merge_stat, current_token)
                    self.update_left_token(input_index, token_index, merge_stat, current_token)
                    self._positions[input_index].replace_pair(token_index, current_token)
                    merge_stat.positions.remove((input_index, token_index))
                merges_done += 1
            return self.itos


        def update_right_token(self, input_index, token_index, merge_stat, current_token):
           position = self._positions[input_index]
           next_token_idx = position.get_next_index(token_index)
           right_token_idx = position.get_second_next_index(token_index)
           if right_token_idx == None:
               return
           pair = (merge_stat.pair[1], position.get_by_index(right_token_idx))
           self.remove_pair_from_stats(merge_stat, pair, input_index, next_token_idx)
           new_pair = (current_token, pair[1])
           self.add_position_to_pair(new_pair, input_index, token_index)

        def update_left_token(self, input_index, token_index, merge_stat, current_token):
           position = self._positions[input_index]
           left_token_idx = position.get_previous_index(token_index)
           if left_token_idx == None:
               return
           pair = ( position.get_by_index(left_token_idx), merge_stat.pair[0])
           self.remove_pair_from_stats(merge_stat, pair, input_index, left_token_idx)
           new_pair = (pair[0] , current_token)
           self.add_position_to_pair(new_pair, input_index, left_token_idx)

        def add_position_to_pair(self, new_pair, input_index, token_index):
            if self._stats.contains(new_pair):
                stat = self._stats.delete_by_map_key(new_pair)
            else:
                stat = StatsEntry(new_pair, set())

            stat.positions.add((input_index, token_index))
            self._stats.push(stat)


        def remove_pair_from_stats(self, merge_stat, pair, input_index, token_index):
           if pair == merge_stat.pair:
               merge_stat.positions.discard((input_index, token_index))
           else:
                if self._stats.contains(pair):
                    stat = self._stats.delete_by_map_key(pair)
                    stat.positions.discard((input_index, token_index))
                    if len(stat.positions) != 0:
                        self._stats.push(stat)


        def update_itos_and_stoi(self, next_token, merge_stat):
           merge_pair = self.itos[merge_stat.pair[0]] + self.itos[merge_stat.pair[1]]
           if merge_pair in self.stoi:
               current_token = self.stoi[merge_pair]
           else:
               current_token = next_token
               self.itos[current_token] = merge_pair
               self.stoi[merge_pair] = current_token
               next_token = current_token+1
           return  (current_token , next_token)

        def get_stats(self):
             self._stats = MaxPriorityMap(
                                heap_key=lambda item: (len(item.positions),tuple(-x for x in item.pair)),
                                map_key=lambda item: item.pair
                            )
             stats={}
             for  text_id in range(len(self.tokens_list)):
                 tokens = self.tokens_list[text_id]
                 for idx in range(len(tokens)-1):
                     pair = (tokens[idx], tokens[idx+1])
                     if pair not in stats:
                         stats[pair]=StatsEntry(pair, set())
                     stats[pair].positions.add((text_id, idx))
             for key in stats.keys():
                self._stats.push(stats[key])

class TokenNode:
    def __init__(self, byte_value, token, parent, children):
        self.parent = parent
        self.byte_value = byte_value
        self.token = token
        self.children = children

class ToTokensConverter:

    def __init__(self, itos):
        self.itos = itos
        self.unk_token=1
        self._init_token_tree()

    def to_tokens(self, strings):
        return [
            self._to_tokens(string)
            for string in strings
        ]

    def _to_tokens(self, string):

        words = string.split()
        tokens = []
        for word in words:
            data = word.encode("utf-8") + b"</w>"
            word_tokens = self._tokenize_word(data)
            tokens.extend(word_tokens)
        return tokens

    def _tokenize_word(self, data):

        tokens = []

        byte_index = 0

        while byte_index < len(data):

            current_nodes = self._token_roots

            last_node_with_token = None
            last_index_with_token = byte_index

            while byte_index < len(data):

                byte_value = data[byte_index]

                if byte_value in current_nodes:

                    node = current_nodes[byte_value]

                    if node.token is not None:
                        last_node_with_token = node
                        last_index_with_token = byte_index

                    current_nodes = node.children
                    byte_index += 1

                else:
                    break

            if last_node_with_token is None:
                tokens.append(self.unk_token)

                # Move forward by one byte
                byte_index += 1

            else:
                tokens.append(last_node_with_token.token)
                byte_index = last_index_with_token + 1

        return tokens

    def _init_token_tree(self):

        self._token_roots = {}

        for token, token_bytes in self.itos.items():

            if token < 4:
                continue

            current_nodes = self._token_roots
            last_node = None

            for byte_value in token_bytes:

                if byte_value in current_nodes:

                    last_node = current_nodes[byte_value]

                else:

                    last_node = TokenNode(
                        byte_value,
                        None,
                        last_node,
                        {}
                    )

                    current_nodes[byte_value] = last_node

                current_nodes = last_node.children

            if last_node.token is not None:
                raise KeyError("The token is already taken")

            last_node.token = token

class BPETokenizer:
    def __init__(self, num_merges):
        self._num_merges = num_merges
        self.itos = {}
    def get_vocab_size(self):
        return len(self.itos)
    def train(self, texts):
        self.create_base_vocab()

        tokens_list = self.get_ids_from_text(texts)
        # for tokens in tokens_list:
        #     print(tokens)
        trainer = TokenizerTrainer(
                        tokens_list,
                        self.itos,
                        self._num_merges
                    )

        trainer.train(261)

        return self.itos, tokens_list

    def encode(self, text):
        converter = ToTokensConverter(self.itos)
        return converter.to_tokens([text])[0]

    def decode(self, tokens):
        return (
                b"".join(
                    self.itos[token]
                    for token in tokens
                )
                .decode("utf-8", errors="replace")
                .replace("</w>", " ")
                .strip()
            )

    def create_base_vocab(self):
        self.itos = {
            0: b"<pad>",
            1: b"<unk>",
            2: b"<s>",
            3: b"</s>",
            260 : b"</w>"
        }
        for i in range(256):
            self.itos[i + 4] = bytes([i])

    def get_ids_from_text(self, texts):
        tokens_list=[]
        for text in texts:
            words = text.split()
            for word in words:
                word_bytes = [byte + 4 for byte in word.encode("utf-8")]
                word_bytes.append(260)
                tokens_list.append(word_bytes)
        return tokens_list
```

```python

import pandas as pd
import numpy as np
from tqdm.auto import tqdm

# ============================================================
# 1. CREATE DATAFRAME
# ============================================================

train_df = pd.DataFrame(train_data).copy()

print("Original records:", len(train_df))


# ============================================================
# 2. REMOVE EXACT DUPLICATE ROWS
# ============================================================

before = len(train_df)

train_df = train_df.drop_duplicates(
    subset=["id", "document", "summary"],
    keep="first"
).reset_index(drop=True)

print("Exact duplicate rows removed:", before - len(train_df))


# ============================================================
# 3. REMOVE EMPTY DOCUMENTS AND SUMMARIES
# ============================================================

empty_document = (
    train_df["document"].isna() |
    train_df["document"].fillna("").str.strip().eq("")
)

empty_summary = (
    train_df["summary"].isna() |
    train_df["summary"].fillna("").str.strip().eq("")
)

print("Empty documents:", empty_document.sum())
print("Empty summaries:", empty_summary.sum())

train_df = train_df.loc[
    ~(empty_document | empty_summary)
].copy().reset_index(drop=True)


# ============================================================
# 4. CALCULATE TOKEN LENGTHS
# ============================================================

def get_token_length(text):
    return len(
        converter.to_tokens([text])[0]
    )

train_df["document_tokens"] = train_df["document"].progress_apply(
    get_token_length
)

train_df["summary_tokens"] = train_df["summary"].progress_apply(
    get_token_length
)


# ============================================================
# 5. FILTER SHORT DOCUMENTS
# ============================================================

MIN_DOCUMENT_TOKENS = 128

short_document = (
    train_df["document_tokens"] < MIN_DOCUMENT_TOKENS
)

print("Short documents removed:", short_document.sum())


# ============================================================
# 6. FILTER LONG SUMMARIES
# ============================================================

MAX_SUMMARY_TOKENS = 6000

long_summary = (
    train_df["summary_tokens"] > MAX_SUMMARY_TOKENS
)

print("Long summaries removed:", long_summary.sum())


# ============================================================
# 7. FILTER SUMMARY/DOCUMENT RATIO
# ============================================================

MAX_SUMMARY_DOCUMENT_RATIO = 0.25

train_df["summary_document_ratio"] = (
    train_df["summary_tokens"] /
    train_df["document_tokens"]
)

long_relative_summary = (
    train_df["summary_document_ratio"] >
    MAX_SUMMARY_DOCUMENT_RATIO
)

print("Summary/document ratio > 0.25:",
      long_relative_summary.sum())


# ============================================================
# 8. APPLY FILTERING
# ============================================================

remove_mask = (
    short_document |
    long_summary |
    long_relative_summary
)

train_clean = (
    train_df.loc[~remove_mask]
    .copy()
    .reset_index(drop=True)
)


# ============================================================
# 9. FINAL SUMMARY
# ============================================================

print("\n========== FINAL DATASET ==========")

print("Records after duplicate removal and empty filtering:",
      len(train_df))

print("Records removed by length filters:",
      remove_mask.sum())

print("Final training records:", len(train_clean))

print("\nDocument token statistics:")
display(train_clean["document_tokens"].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
).to_frame())

print("\nSummary token statistics:")
display(train_clean["summary_tokens"].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
).to_frame())

print("\nSummary/document ratio statistics:")
display(train_clean["summary_document_ratio"].describe())
```

```python

import json
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader

MAX_CHUNK_LEN = 1024
CHUNK_OVERLAP = 128
MAX_CHUNKS = 20

PAD_ID = 0
UNK_ID = 1
BOS_ID = 2
EOS_ID = 3


class LegalSummarizationDataset(Dataset):

    def __init__(
        self,
        json_path,
        converter,
        max_chunk_len=MAX_CHUNK_LEN,
        overlap=CHUNK_OVERLAP,
        max_chunks=MAX_CHUNKS,
        has_labels=True,
    ):

        with open(json_path, "r", encoding="utf-8") as f:
            self.data = json.load(f)

        self.converter = converter
        self.max_chunk_len = max_chunk_len
        self.overlap = overlap
        self.max_chunks = max_chunks
        self.has_labels = has_labels

        assert 0 <= overlap < max_chunk_len - 2
        assert max_chunk_len >= 3

    def __len__(self):
        return len(self.data)

    def tokenize(self, text):
        return self.converter.to_tokens([text])[0]

    def create_chunks(self, token_ids):

        payload_len = self.max_chunk_len - 2
        step = payload_len - self.overlap

        chunks = []

        for start in range(0, len(token_ids), step):

            content = token_ids[start:start + payload_len]

            if not content:
                break

            chunk = [BOS_ID] + content + [EOS_ID]
            chunks.append(chunk)

            if start + payload_len >= len(token_ids):
                break

        if not chunks:
            chunks = [[BOS_ID, EOS_ID]]

        # Select chunks distributed across the document
        if len(chunks) > self.max_chunks:

            indices = np.linspace(
                0,
                len(chunks) - 1,
                self.max_chunks,
                dtype=int
            )

            chunks = [chunks[i] for i in indices]

        return chunks

    def __getitem__(self, idx):

        item = self.data[idx]

        document_ids = self.tokenize(item["document"])

        chunks = self.create_chunks(document_ids)

        summary_ids = None

        if self.has_labels:

            summary_tokens = self.tokenize(item["summary"])

            summary_ids = (
                [BOS_ID] + summary_tokens + [EOS_ID]
            )

        return {
            "id": item["id"],
            "chunks": chunks,
            "num_chunks": len(chunks),
            "document_tokens": len(document_ids),
            "summary_ids": summary_ids,
        }
```

```python

def summarization_collate_fn(batch):

    batch_size = len(batch)

    pad_id = PAD_ID

    # ========================================================
    # 1. PAD DOCUMENT CHUNKS
    # ========================================================

    input_ids = torch.full(
        (batch_size, MAX_CHUNKS, MAX_CHUNK_LEN),
        pad_id,
        dtype=torch.long
    )

    attention_mask = torch.zeros(
        (batch_size, MAX_CHUNKS, MAX_CHUNK_LEN),
        dtype=torch.long
    )

    chunk_mask = torch.zeros(
        (batch_size, MAX_CHUNKS),
        dtype=torch.bool
    )

    for i, item in enumerate(batch):

        for j, chunk in enumerate(item["chunks"]):

            chunk_len = len(chunk)

            input_ids[i, j, :chunk_len] = torch.tensor(
                chunk,
                dtype=torch.long
            )

            attention_mask[i, j, :chunk_len] = 1

            chunk_mask[i, j] = True

    # ========================================================
    # 2. PAD REFERENCE SUMMARIES
    # ========================================================

    summary_ids = None
    summary_mask = None

    if batch[0]["summary_ids"] is not None:

        max_summary_len = max(
            len(item["summary_ids"])
            for item in batch
        )

        summary_ids = torch.full(
            (batch_size, max_summary_len),
            pad_id,
            dtype=torch.long
        )

        summary_mask = torch.zeros(
            (batch_size, max_summary_len),
            dtype=torch.long
        )

        for i, item in enumerate(batch):

            ids = item["summary_ids"]
            summary_len = len(ids)

            summary_ids[i, :summary_len] = torch.tensor(
                ids,
                dtype=torch.long
            )

            summary_mask[i, :summary_len] = 1

    # ========================================================
    # 3. RETURN BATCH
    # ========================================================

    return {
        "ids": [item["id"] for item in batch],

        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "chunk_mask": chunk_mask,

        "summary_ids": summary_ids,
        "summary_mask": summary_mask,

        "num_chunks": torch.tensor(
            [item["num_chunks"] for item in batch],
            dtype=torch.long
        ),

        "document_tokens": torch.tensor(
            [item["document_tokens"] for item in batch],
            dtype=torch.long
        )
    }
```

```python
import math
import torch
import torch.nn as nn
import torch.nn.functional as F


def chunked_gqa_linear_attn(q, k, v, chunk=128, eps=1e-6, padding_mask = None):
    """
    q: (B, Hq, T, F)   k: (B, Hkv, T, F)   v: (B, Hkv, T, D)   all float32, features >= 0
    Replaces the kv / cumsum / einsum block. Returns out (B, Hq, T, D) and final state (S, z).
    """
    B, Hq, T, Fd = q.shape
    Hkv, D = k.shape[1], v.shape[-1]
    R = Hq // Hkv
    if padding_mask is not None:
        valid = (~padding_mask.bool())[:, None, :, None]
        k = k * valid.to(k.dtype)
        v = v * valid.to(v.dtype)
    q = q.view(B, Hkv, R, T, Fd)                      # query head h uses KV head h // R

    S = q.new_zeros(B, Hkv, Fd, D)                    # running sum of phi(k)^T v
    z = q.new_zeros(B, Hkv, Fd)                       # running sum of phi(k)
    tri = torch.tril(torch.ones(chunk, chunk, dtype=torch.bool, device=q.device))

    outs = []
    for s in range(0, T, chunk):
        qc = q[:, :, :, s:s + chunk]                  # (B,Hkv,R,c,F)
        kc = k[:, :, s:s + chunk]                     # (B,Hkv,c,F)
        vc = v[:, :, s:s + chunk]                     # (B,Hkv,c,D)
        c = qc.shape[3]

        a = (qc @ kc.unsqueeze(2).transpose(-1, -2)).masked_fill(~tri[:c, :c], 0.0)  # inside chunk
        num = qc @ S.unsqueeze(2) + a @ vc.unsqueeze(2)                             # past + current
        den = (qc * z[:, :, None, None, :]).sum(-1, keepdim=True) + a.sum(-1, keepdim=True)
        outs.append(num / (den + eps))

        S = S + kc.transpose(-1, -2) @ vc             # update state after the chunk
        z = z + kc.sum(dim=2)

    return torch.cat(outs, dim=3).reshape(B, Hq, T, D), (S, z)


class LinearAttention(nn.Module):
    def __init__(self, d_model, n_heads, n_kv_heads, dropout=0.0, feat_dim=32, chunk=128):
        super().__init__()
        assert d_model % n_heads == 0 and n_heads % n_kv_heads == 0
        self.n_heads, self.n_kv_heads = n_heads, n_kv_heads
        self.n_rep = n_heads // n_kv_heads
        self.d_head = d_model // n_heads
        self.chunk = chunk
        self.feat_dim = feat_dim
        self.q_proj = nn.Linear(d_model, n_heads * self.d_head, bias=False)
        self.k_proj = nn.Linear(d_model, n_kv_heads * self.d_head, bias=False)
        self.v_proj = nn.Linear(d_model, n_kv_heads * self.d_head, bias=False)
        self.out_proj = nn.Linear(n_heads * self.d_head, d_model, bias=False)

        # kernel: phi(x) = (gamma * norm(x) + beta)^2  (gamma, beta in the LayerNorm)
        self.q_kernel = nn.Linear(self.d_head, feat_dim, bias=True)
        self.k_kernel = nn.Linear(self.d_head, feat_dim, bias=True)
        self.q_norm = nn.LayerNorm(feat_dim)
        self.k_norm = nn.LayerNorm(feat_dim)

        self.resid_dropout = nn.Dropout(dropout)

    def _project(self, x):
        B, T, _ = x.shape
        q = self.q_proj(x).view(B, T, self.n_heads, self.d_head).transpose(1, 2)
        k = self.k_proj(x).view(B, T, self.n_kv_heads, self.d_head).transpose(1, 2)
        v = self.v_proj(x).view(B, T, self.n_kv_heads, self.d_head).transpose(1, 2)
        q_phi = self.q_norm(self.q_kernel(q)).float() ** 2     # (B,n_heads,T,F)
        k_phi = self.k_norm(self.k_kernel(k)).float() ** 2     # (B,n_kv,T,F) 
        return q_phi, k_phi, v.float()

    def forward(self, x, padding_mask=None, return_state=False):
        B, T, C = x.shape
        q_phi, k_phi, v = self._project(x)
        with torch.autocast(device_type="cuda", enabled=False):
            out, state = chunked_gqa_linear_attn(q_phi, k_phi, v, self.chunk, padding_mask=padding_mask)
        out = F.rms_norm(out.to(x.dtype), (self.d_head,))
        out = self.out_proj(out.transpose(1, 2).reshape(B, T, C))
        out = self.resid_dropout(out)
        return (out, state) if return_state else out

    @torch.no_grad()
    def step(self, x_t, S, z, eps=1e-6):
        """One-token decoding. x_t: (B,1,C), S: (B,Hkv,F,D), z: (B,Hkv,F), both float32."""
        B = x_t.shape[0]
        q_phi, k_phi, v = self._project(x_t)
        S = S + k_phi.transpose(-1, -2) @ v
        z = z + k_phi.squeeze(2)
    
        qg = q_phi.reshape(
            B, self.n_kv_heads, self.n_rep, self.feat_dim
        )
    
        numerator = torch.einsum("bhrf,bhfd->bhrd", qg, S)
        denominator = (qg * z.unsqueeze(2)).sum(-1, keepdim=True)
    
        out = numerator / (denominator + eps)
        out = F.rms_norm(out.to(x_t.dtype), (self.d_head,))
        out = out.reshape(B, 1, self.n_heads * self.d_head)
        out = self.out_proj(out)
    
        return out, S, z
```

```python

"""Decoder-only Transformer LM -- Block now passes use_rope through to attention, and
DecoderLM only builds/adds the sinusoidal table when pos_emb_type == "sinusoidal"
(everything else unchanged from Question 2)."""
import math
from dataclasses import dataclass
import torch
import torch.nn as nn
import torch.nn.functional as F



@dataclass
class LMConfig:
    vocab_size: int
    d_model: int = 256
    n_layers: int = 4
    n_heads: int = 4
    n_kv_heads: int = 4
    d_ff: int = 1024
    max_seq_len: int = 256
    dropout: float = 0.1
    pos_emb_type: str = "sinusoidal"   # "sinusoidal" or "rope"
    feat_dim: int = 32                 # kernel feature dimension F
    chunk: int = 128 


def sinusoidal_positional_encoding(max_len: int, d_model: int) -> torch.Tensor:
    PE = torch.zeros(max_len, d_model)
    for pos in range(max_len):
      for i in range(d_model//2):
        PE[pos, 2*i] = math.sin(pos / 10000**(2*i/d_model))
        PE[pos, 2*i+1] = math.cos(pos / 10000**(2*i/d_model))
    return PE


class FeedForward(nn.Module):
    def __init__(self, d_model: int, d_ff: int, dropout: float):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.GELU(),
            nn.Linear(d_ff, d_model),
            nn.Dropout(dropout)
        )   # unchanged from Question 1

    def forward(self, x):
        return self.net(x)



class Block(nn.Module):
    def __init__(self, cfg: LMConfig):
        super().__init__()
        self.ln1 = nn.LayerNorm(cfg.d_model)
        self.attn = LinearAttention(cfg.d_model, cfg.n_heads, cfg.n_kv_heads,
                                       cfg.dropout, cfg.feat_dim, cfg.chunk)
        self.ln2 = nn.LayerNorm(cfg.d_model)
        self.ffn = FeedForward(cfg.d_model, cfg.d_ff, cfg.dropout)

    def forward(self, x, padding_mask=None, return_state=False):
            if return_state:
                a, state = self.attn(
                    self.ln1(x),
                    padding_mask=padding_mask,
                    return_state=True,
                )
            else:
                a = self.attn(
                    self.ln1(x),
                    padding_mask=padding_mask,
                )
                state = None
        
            x = x + a
            x = x + self.ffn(self.ln2(x))
        
            return (x, state) if return_state else x

    @torch.no_grad()
    def step(self, x_t, state):
        S, z = state
        a, S, z = self.attn.step(self.ln1(x_t), S, z)
        x = x_t + a
        x = x + self.ffn(self.ln2(x))
        return x, (S, z)

def sinusoidal_positional_encoding(max_len, d_model):
    pos = torch.arange(max_len, dtype=torch.float32)[:, None]
    i = torch.arange(d_model // 2, dtype=torch.float32)[None, :]
    ang = pos / (10000 ** (2 * i / d_model))
    pe = torch.zeros(max_len, d_model)
    pe[:, 0::2], pe[:, 1::2] = torch.sin(ang), torch.cos(ang)
    return pe


class DecoderLM(nn.Module):
    def __init__(self, cfg: LMConfig):
        super().__init__()
        self.cfg = cfg
        self.tok_emb = nn.Embedding(cfg.vocab_size, cfg.d_model)
        if cfg.pos_emb_type == "sinusoidal":
            self.register_buffer("pos_emb", sinusoidal_positional_encoding(cfg.max_seq_len, cfg.d_model),
                                 persistent=False)
        self.blocks = nn.ModuleList([Block(cfg) for _ in range(cfg.n_layers)])
        self.ln_f = nn.LayerNorm(cfg.d_model)
        self.head = nn.Linear(cfg.d_model, cfg.vocab_size, bias=False)
        self.head.weight = self.tok_emb.weight
        with torch.no_grad():
                nn.init.uniform_(
                    self.tok_emb.weight,
                    -1 / math.sqrt(cfg.d_model),
                    1 / math.sqrt(cfg.d_model)
                )

    def forward(self, idx, padding_mask=None):
        x = self.tok_emb(idx)
    
        if self.cfg.pos_emb_type == "sinusoidal":
            T = idx.shape[1]
            x = x + self.pos_emb[:T].unsqueeze(0)
    
        for block in self.blocks:
            x = block(x, padding_mask=padding_mask)

        return self.head(self.ln_f(x))

    # ---------- inference ----------
    @torch.no_grad()
    def prefill(self, idx):
            x = self.tok_emb(idx)

            if self.cfg.pos_emb_type == "sinusoidal":
                T = idx.shape[1]
                x = x + self.pos_emb[:T].unsqueeze(0)
        
            states = []
        
            for block in self.blocks:
                x, state = block(x, return_state=True)
                states.append(state)
        
            logits = self.head(self.ln_f(x[:, -1:]))

            return logits, states          # logits of last token

    @torch.no_grad()
    def step(self, tok, pos, states):
        x = self.tok_emb(tok)

        if self.cfg.pos_emb_type == "sinusoidal":
            if pos >= self.cfg.max_seq_len:
                raise ValueError("Maximum sequence length exceeded")
    
        x = x + self.pos_emb[pos:pos + 1].unsqueeze(0)

        new_states = []
    
        for block, state in zip(self.blocks, states):
    
            x, new_state = block.step(x, state)
    
            new_states.append(new_state)
    
        logits = self.head(self.ln_f(x))

        return logits, new_states
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class LoraLinear(nn.Module):

    def __init__(self, layer, rank=16, alpha=32, dropout=0.05):
        super().__init__()

        self.layer = layer

        for param in self.layer.parameters():
            param.requires_grad = False

        self.in_features = layer.in_features
        self.out_features = layer.out_features

        self.rank = rank
        self.alpha = alpha
        self.scaling = alpha / rank

        self.dropout = nn.Dropout(dropout)

        # A: [rank, input_features]
        self.A = nn.Parameter(
            torch.empty(rank, self.in_features)
        )

        # B: [output_features, rank]
        self.B = nn.Parameter(
            torch.zeros(self.out_features, rank)
        )

        nn.init.kaiming_uniform_(
            self.A,
            a=5 ** 0.5
        )

    def forward(self, x):

        original_output = self.layer(x)

        lora_output = F.linear(
            self.dropout(x),
            self.A
        )

        lora_output = F.linear(
            lora_output,
            self.B
        )

        return original_output + self.scaling * lora_output
```

```python
class LoraDecoderLM(nn.Module):

    def __init__(
        self,
        base_model,
        rank=16,
        alpha=32,
        lora_dropout=0.05,
        projections=("q_proj", "k_proj", "v_proj", "out_proj")
    ):
        super().__init__()

        self.model = base_model

        # Freeze the pretrained model
        for param in self.model.parameters():
            param.requires_grad = False

        self.rank = rank
        self.alpha = alpha
        self.projections = projections

        self._add_lora(
            rank=rank,
            alpha=alpha,
            dropout=lora_dropout,
            projections=projections
        )

    def _add_lora(self, rank, alpha, dropout, projections):

        for block in self.model.blocks:

            attention = block.attn

            for proj_name in projections:

                layer = getattr(attention, proj_name)

                if not isinstance(layer, nn.Linear):
                    raise TypeError(
                        f"{proj_name} is not nn.Linear: {type(layer)}"
                    )

                setattr(
                    attention,
                    proj_name,
                    LoraLinear(
                        layer=layer,
                        rank=rank,
                        alpha=alpha,
                        dropout=dropout
                    )
                )

    def forward(self, input_ids, padding_mask=None):

        return self.model(
            idx=input_ids,
            padding_mask=padding_mask
        )

    # Linear-attention inference methods
    @torch.no_grad()
    def prefill(self, input_ids):

        return self.model.prefill(input_ids)

    @torch.no_grad()
    def step(self, token, position, states):

        return self.model.step(token, position, states)

    def get_trainable_parameters(self):

        return [
            p for p in self.parameters()
            if p.requires_grad
        ]

    def print_trainable_parameters(self):

        trainable = sum(
            p.numel() for p in self.parameters()
            if p.requires_grad
        )

        total = sum(p.numel() for p in self.parameters())

        print(f"Trainable parameters: {trainable:,}")
        print(f"Total parameters:     {total:,}")
        print(f"Trainable percentage: {100 * trainable / total:.4f}%")
```

```python
import re
import json
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F

from tqdm.auto import tqdm

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Special token IDs from your custom tokenizer
PAD_ID = 0
UNK_ID = 1
BOS_ID = 2
EOS_ID = 3

# Chunk configuration (same as your existing dataset)
MAX_CHUNK_LEN = 1024
CHUNK_OVERLAP = 128
MAX_CHUNKS = 20

PAYLOAD_LEN = MAX_CHUNK_LEN - 2
STRIDE = PAYLOAD_LEN - CHUNK_OVERLAP

print("Device:", DEVICE)
print("Payload length:", PAYLOAD_LEN)
print("Stride:", STRIDE)
```

```python
def split_summary_into_sentences(text):
    text = str(text).strip()

    if not text:
        return []

    sentences = re.split(r'(?<=[.!?])\s+', text)

    return [
        sentence.strip()
        for sentence in sentences
        if sentence.strip()
    ]
```

```python
model_phase1 = model_phase1.to(DEVICE)
model_phase1.eval()

@torch.no_grad()
def get_text_embedding(token_ids):
    """
    Generate one embedding for a text sequence using
    the Phase 1 Decoder's final hidden representations.
    """

    token_ids = token_ids[:PAYLOAD_LEN]

    if len(token_ids) == 0:
        return None

    input_ids = torch.tensor(
        [[BOS_ID] + token_ids + [EOS_ID]],
        dtype=torch.long,
        device=DEVICE
    )

    # Token embeddings
    x = model_phase1.tok_emb(input_ids)

    # Decoder blocks
    for block in model_phase1.blocks:
        x, _ = block(x)

    # Final layer normalization
    hidden = model_phase1.ln_f(x)

    # Exclude BOS and EOS
    token_hidden = hidden[:, 1:-1, :]

    # Mean pooling
    embedding = token_hidden.mean(dim=1)

    # Normalize for cosine similarity
    embedding = F.normalize(embedding, p=2, dim=-1)

    return embedding.squeeze(0)
```

```python
@torch.no_grad()
def align_summary_to_chunks(document, summary, dataset_item):
    """
    Assign each summary sentence to its most similar document chunk.

    dataset_item is obtained from your existing train_dataset,
    ensuring we reuse exactly the same chunking and chunk selection.
    """

    chunks = dataset_item["chunks"]

    sentences = split_summary_into_sentences(summary)

    if not sentences:
        return [""] * len(chunks), []

    # --------------------------------------------------
    # 1. Get document chunk embeddings
    # --------------------------------------------------

    chunk_embeddings = []

    for chunk in chunks:
        # Remove BOS and EOS before embedding
        content_ids = chunk[1:-1]

        embedding = get_text_embedding(content_ids)

        if embedding is not None:
            chunk_embeddings.append(embedding)

    if not chunk_embeddings:
        return [""] * len(chunks), []

    chunk_embeddings = torch.stack(chunk_embeddings)

    # --------------------------------------------------
    # 2. Get summary sentence embeddings
    # --------------------------------------------------

    sentence_embeddings = []
    valid_sentences = []

    for sentence in sentences:

        sentence_ids = converter.to_tokens([sentence])[0]

        # Very long sentences are truncated to the model's
        # available payload length.
        embedding = get_text_embedding(sentence_ids)

        if embedding is not None:
            sentence_embeddings.append(embedding)
            valid_sentences.append(sentence)

    if not sentence_embeddings:
        return [""] * len(chunks), []

    sentence_embeddings = torch.stack(sentence_embeddings)

    # --------------------------------------------------
    # 3. Cosine similarity
    # --------------------------------------------------

    similarity = sentence_embeddings @ chunk_embeddings.T

    # --------------------------------------------------
    # 4. Assign each sentence to its best chunk
    # --------------------------------------------------

    best_chunks = similarity.argmax(dim=1).cpu().tolist()
    best_scores = similarity.max(dim=1).values.cpu().tolist()

    aligned_targets = [""] * len(chunks)

    for sentence, chunk_idx in zip(valid_sentences, best_chunks):

        if aligned_targets[chunk_idx]:
            aligned_targets[chunk_idx] += " "

        aligned_targets[chunk_idx] += sentence

    alignment_info = [
        {
            "sentence": sentence,
            "chunk_index": int(chunk_idx),
            "similarity": float(score)
        }
        for sentence, chunk_idx, score
        in zip(valid_sentences, best_chunks, best_scores)
    ]

    return aligned_targets, alignment_info
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence
from tqdm.auto import tqdm

PAD_ID = 0
BOS_ID = 2
EOS_ID = 3

MAX_SEQ_LEN = 1024
MAX_SOURCE_LEN = 768
BATCH_SIZE = 2
class AlignedSummarizationDataset(Dataset):

    def __init__(self, aligned_data):

        self.examples = []

        self.total_target_tokens = 0
        self.retained_target_tokens = 0

        for doc in aligned_data:

            for chunk, target in zip(
                doc["chunks"],
                doc["chunk_targets"]
            ):

                if not target.strip():
                    continue

                # Document chunk
                source_ids = list(chunk)

                # Keep source within the limit
                if len(source_ids) > MAX_SOURCE_LEN:
                    source_ids = (
                        source_ids[:MAX_SOURCE_LEN - 1]
                        + [EOS_ID]
                    )

                # Complete target summary
                target_ids = converter.to_tokens([target])[0]

                self.total_target_tokens += len(target_ids)

                # Available target space
                target_budget = (
                    MAX_SEQ_LEN - len(source_ids) - 1
                )

                if target_budget <= 0:
                    continue

                # Split target into smaller pieces
                for start in range(0, len(target_ids), target_budget):

                    target_part = target_ids[
                        start:start + target_budget
                    ]

                    is_last = (
                        start + target_budget >= len(target_ids)
                    )

                    # Source + target
                    input_ids = source_ids + target_part

                    # Ignore source tokens in loss
                    labels = (
                        [-100] * len(source_ids)
                        + target_part
                    )

                    # EOS only for final target segment
                    if is_last:
                        input_ids.append(EOS_ID)
                        labels.append(EOS_ID)

                    self.examples.append({
                        "input_ids": input_ids,
                        "labels": labels
                    })

                    self.retained_target_tokens += len(target_part)

        print("Total training examples:", len(self.examples))
        print("Original target tokens:", self.total_target_tokens)
        print("Retained target tokens:", self.retained_target_tokens)
        print(
            "Retention:",
            round(
                100 * self.retained_target_tokens /
                max(self.total_target_tokens, 1),
                2
            ),
            "%"
        )

    def __len__(self):
        return len(self.examples)

    def __getitem__(self, idx):
        return self.examples[idx]
```

```python
def aligned_collate_fn(batch):

    input_ids = [
        torch.tensor(x["input_ids"], dtype=torch.long)
        for x in batch
    ]

    labels = [
        torch.tensor(x["labels"], dtype=torch.long)
        for x in batch
    ]

    input_ids = pad_sequence(
        input_ids,
        batch_first=True,
        padding_value=PAD_ID
    )

    labels = pad_sequence(
        labels,
        batch_first=True,
        padding_value=-100
    )

    attention_mask = (input_ids != PAD_ID).long()

    return {
        "input_ids": input_ids,
        "labels": labels,
        "attention_mask": attention_mask
    }
```

```python
import torch
import torch.nn.functional as F
from tqdm.auto import tqdm


def train_one_epoch(
    model,
    loader,
    optimizer,
    device,
    scaler=None,
    use_amp=True
):

    model.train()

    total_loss = 0.0
    total_tokens = 0

    progress = tqdm(loader, desc="Training")

    trainable_params = [
        p for p in model.parameters()
        if p.requires_grad
    ]

    for batch in progress:

        input_ids = batch["input_ids"].to(device)
        labels = batch["labels"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        # True means padding
        padding_mask = ~attention_mask.bool()

        optimizer.zero_grad(set_to_none=True)

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=use_amp and torch.cuda.is_available()
        ):

            logits = model(
                input_ids=input_ids,
                padding_mask=padding_mask
            )

            shift_logits = logits[:, :-1, :].contiguous()
            shift_labels = labels[:, 1:].contiguous()

            loss_sum = F.cross_entropy(
                shift_logits.reshape(-1, shift_logits.size(-1)),
                shift_labels.reshape(-1),
                ignore_index=-100,
                reduction="sum"
            )

            valid_tokens = (shift_labels != -100).sum()
            loss = loss_sum / valid_tokens.clamp_min(1)

        if scaler is not None and scaler.is_enabled():

            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)

            torch.nn.utils.clip_grad_norm_(
                trainable_params,
                max_norm=1.0
            )

            scaler.step(optimizer)
            scaler.update()

        else:

            loss.backward()

            torch.nn.utils.clip_grad_norm_(
                trainable_params,
                max_norm=1.0
            )

            optimizer.step()

        total_loss += loss_sum.detach().item()
        total_tokens += valid_tokens.item()

        progress.set_postfix(
            loss=f"{loss.item():.4f}"
        )

    return total_loss / max(total_tokens, 1)
```

```python
@torch.no_grad()
def evaluate(
    model,
    loader,
    device,
    use_amp=True
):

    model.eval()

    total_loss = 0.0
    total_tokens = 0

    progress = tqdm(loader, desc="Validation")

    for batch in progress:

        input_ids = batch["input_ids"].to(device)
        labels = batch["labels"].to(device)
        attention_mask = batch["attention_mask"].to(device)

        padding_mask = ~attention_mask.bool()

        with torch.autocast(
            device_type="cuda",
            dtype=torch.float16,
            enabled=use_amp and torch.cuda.is_available()
        ):

            logits = model(
                input_ids=input_ids,
                padding_mask=padding_mask
            )

            shift_logits = logits[:, :-1, :].contiguous()
            shift_labels = labels[:, 1:].contiguous()

            loss_sum = F.cross_entropy(
                shift_logits.reshape(-1, shift_logits.size(-1)),
                shift_labels.reshape(-1),
                ignore_index=-100,
                reduction="sum"
            )

            valid_tokens = (shift_labels != -100).sum()

        total_loss += loss_sum.item()
        total_tokens += valid_tokens.item()

        progress.set_postfix(
            val_loss=f"{(loss_sum / valid_tokens.clamp_min(1)).item():.4f}"
        )

    return total_loss / max(total_tokens, 1)
```

```python
import numpy as np
import torch

PAD_ID = 0
BOS_ID = 2
EOS_ID = 3

MAX_SEQ_LEN = 1024
MAX_SOURCE_LEN = 768

CHUNK_OVERLAP = 128
MAX_CHUNKS = 20


def split_document_into_chunks(
    document,
    max_chunks=20
):

    token_ids = converter.to_tokens([document])[0]

    payload_len = MAX_SOURCE_LEN - 2
    stride = payload_len - CHUNK_OVERLAP

    chunks = []

    for start in range(0, len(token_ids), stride):

        content = token_ids[start:start + payload_len]

        if not content:
            break

        chunk = [BOS_ID] + content + [EOS_ID]

        chunks.append(chunk)

        if start + payload_len >= len(token_ids):
            break

    if len(chunks) > max_chunks:

        indices = np.linspace(
            0,
            len(chunks) - 1,
            max_chunks,
            dtype=int
        )

        chunks = [chunks[i] for i in indices]

    return chunks
```

```python
import torch
import torch.nn.functional as F


def sample_next_token(
    logits,
    context_ids,
    temperature=0.7,
    top_k=40,
    repetition_penalty=1.2,
    no_repeat_ngram_size=3
):

    logits = logits.clone()

    # Repetition penalty
    if repetition_penalty != 1.0:

        for token_id in set(context_ids):

            if logits[0, token_id] > 0:
                logits[0, token_id] /= repetition_penalty
            else:
                logits[0, token_id] *= repetition_penalty

    # No-repeat n-gram blocking
    n = no_repeat_ngram_size

    if n > 0 and len(context_ids) >= n - 1:

        prefix = context_ids[-(n - 1):]

        for i in range(len(context_ids) - n + 1):

            if context_ids[i:i + n - 1] == prefix:

                banned_token = context_ids[i + n - 1]

                logits[0, banned_token] = float("-inf")

    # Temperature
    logits = logits / temperature

    # Top-k filtering
    if top_k is not None and top_k > 0:

        k = min(top_k, logits.size(-1))

        values, _ = torch.topk(logits, k=k, dim=-1)

        threshold = values[:, -1].unsqueeze(-1)

        logits = logits.masked_fill(
            logits < threshold,
            float("-inf")
        )

    probabilities = F.softmax(logits, dim=-1)

    next_token = torch.multinomial(
        probabilities,
        num_samples=1
    )

    return next_token
```

```python
@torch.no_grad()
def generate_summary_linear(
    model,
    chunk,
    max_new_tokens=200,
    temperature=0.7,
    top_k=40,
    repetition_penalty=1.2,
    no_repeat_ngram_size=3,
    device=DEVICE
):

    model.eval()

    # Unwrap DataParallel
    if isinstance(model, torch.nn.DataParallel):
        model = model.module

    # Convert prompt to tensor
    input_ids = torch.tensor(
        [chunk],
        dtype=torch.long,
        device=device
    )

    prompt_length = input_ids.size(1)

    # Initial context
    context_ids = list(chunk)

    generated_ids = []

    # ------------------------------------------------
    # 1. PREFILL
    # ------------------------------------------------

    logits, states = model.prefill(input_ids)

    # ------------------------------------------------
    # 2. AUTOREGRESSIVE GENERATION
    # ------------------------------------------------

    for step in range(max_new_tokens):

        # Check context limit before generating
        next_position = prompt_length + len(generated_ids)

        if next_position >= MAX_SEQ_LEN:
            break

        next_logits = logits[:, -1, :]

        # Sampling
        next_token = sample_next_token(
            logits=next_logits,
            context_ids=context_ids,
            temperature=temperature,
            top_k=top_k,
            repetition_penalty=repetition_penalty,
            no_repeat_ngram_size=no_repeat_ngram_size
        )

        token_id = next_token.item()

        # Stop at EOS
        if token_id == EOS_ID:
            break

        generated_ids.append(token_id)
        context_ids.append(token_id)

        # ------------------------------------------------
        # 3. INCREMENTAL LINEAR ATTENTION STEP
        # ------------------------------------------------

        logits, states = model.step(
            token=next_token,
            position=next_position,
            states=states
        )

    # Decode
    summary = tokenizer.decode(generated_ids)

    return summary
```

```python
def generate_document_summary_linear(
    model,
    document,
    max_chunks=20,
    max_new_tokens=200
):

    chunks = split_document_into_chunks(
        document,
        max_chunks=max_chunks
    )

    chunk_summaries = []

    print("Total chunks:", len(chunks))

    for i, chunk in enumerate(chunks):

        summary = generate_summary_linear(
            model=model,
            chunk=chunk,
            max_new_tokens=max_new_tokens,
            temperature=0.7,
            top_k=40,
            repetition_penalty=1.2,
            no_repeat_ngram_size=3
        )

        chunk_summaries.append(summary)

        generated_length = len(
            converter.to_tokens([summary])[0]
        )

        print(
            f"Chunk {i+1}/{len(chunks)} | "
            f"Generated tokens: {generated_length}"
        )

    return chunk_summaries
```

```python
import json
from tqdm.auto import tqdm


def generate_all_predictions_linear(
    model,
    test_data,
    output_path="linear_attention_lora_predictions.json",
    max_chunks=20,
    max_new_tokens=200
):

    predictions = []

    for item in tqdm(test_data, desc="Generating predictions"):

        doc_id = str(item["id"])
        document = item["document"]

        chunk_summaries = generate_document_summary_linear(
            model=model,
            document=document,
            max_chunks=max_chunks,
            max_new_tokens=max_new_tokens
        )

        final_summary = " ".join(
            summary.strip()
            for summary in chunk_summaries
            if summary.strip()
        )

        predictions.append({
            "id": doc_id,
            "summary": final_summary
        })

    with open(output_path, "w", encoding="utf-8") as f:

        json.dump(
            predictions,
            f,
            ensure_ascii=False,
            indent=2
        )

    print("\nGeneration completed!")
    print("Total predictions:", len(predictions))
    print("Saved to:", output_path)

    return predictions
```
